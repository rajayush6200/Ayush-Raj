# 3. Handling Missing Data

This section examines the dataset for missing or null values.

```python
# Check the number of missing values in each column

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Calculate the total number of missing values

total_missing = missing_values.sum()

print(f"Total Missing Values in the Dataset: {total_missing}")
```

```python
# Calculate the percentage of missing values

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage (%)": missing_percentage
})

missing_summary
```

```python
# Visualize missing values using a heatmap

plt.figure(figsize=(14, 6))
sns.heatmap(df.isnull(), cbar=False, cmap="viridis")

plt.title("Missing Values Heatmap")
plt.xlabel("Columns")
plt.ylabel("Rows")

plt.show()
```

```python
# Handle missing values (if present)

if total_missing == 0:
    print("No missing values found in the dataset.")
    print("No data imputation or removal is required.")
else:
    df.fillna(df.median(numeric_only=True), inplace=True)
    print("Missing values have been replaced using the median.")
```

```python
# Verify that no missing values remain

remaining_missing = df.isnull().sum().sum()

print(f"Remaining Missing Values: {remaining_missing}")
```

### Observation

- The dataset was examined for missing values using null value counts, percentage analysis, and a heatmap.
- Missing values were identified in multiple numerical columns of the dataset.
- The missing values were handled by replacing them with the median value of their respective numerical columns.
- After preprocessing, the dataset was verified again and no missing values remained.
- The dataset is now complete and ready for the subsequent data cleaning and transformation steps.

# 7. Bivariate Analysis

This section examines the relationship between two numerical variables in the dataset.

```python
# Scatter plot of Population vs Murders

plt.figure(figsize=(8,5))

plt.scatter(
    df["popul"],
    df["murders"],
    alpha=0.6
)

plt.title("Population vs Number of Murders")
plt.xlabel("Population")
plt.ylabel("Murders")

plt.show()
```

### Observation

- The scatter plot shows a positive relationship between population and the number of murders.
- Counties with larger populations generally tend to report a higher number of murders.
- Although an increasing trend is visible, the spread of the points indicates that factors other than population also influence murder counts.
- A few counties with very large populations appear as high-value observations in the upper-right region of the plot.

```python
# Regression plot of Population Density vs Murder Rate

plt.figure(figsize=(8,5))

sns.regplot(
    x="density",
    y="murdrate",
    data=df,
    scatter_kws={"alpha":0.4},
    line_kws={"color":"red"}
)

plt.title("Population Density vs Murder Rate")
plt.xlabel("Population Density")
plt.ylabel("Murder Rate")

plt.show()
```

### Observation

- The regression plot indicates a weak positive relationship between population density and murder rate.
- Although the regression line shows a slight upward trend, the data points are widely scattered, indicating a weak linear association.
- Most counties have low population density and relatively low murder rates, while a few observations show exceptionally high murder rates.
- This suggests that population density alone is not a strong predictor of murder rate.

```python
# Correlation heatmap

plt.figure(figsize=(12,10))

correlation_matrix = df.corr(numeric_only=True)

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    annot=False,
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

- The correlation heatmap provides an overview of the relationships among all numerical variables.
- Strong positive correlations are visible between a few pairs of variables, while many other variables exhibit weak or moderate correlations.
- The heatmap helps identify variables that move together and those that have little or no linear relationship.

### Overall Observation

- Bivariate analysis was performed using a scatter plot, regression plot, and correlation heatmap.
- The scatter plot showed a positive relationship between population and the number of murders.
- The regression plot indicated a weak positive relationship between population density and murder rate.
- The correlation heatmap summarized the relationships among all numerical variables and highlighted both strong and weak correlations.
- These visualizations provide valuable insights into how different variables are associated and support further multivariate analysis.

# 8. Multivariate Analysis

This section explores the relationships among multiple numerical variables simultaneously using advanced visualization techniques.

```python
# Pair Plot for Selected Variables

selected_columns = ["popul", "murders", "murdrate", "density"]

sns.pairplot(
    df[selected_columns],
    diag_kind="kde"
)

plt.show()
```

### Observation

- The pair plot provides pairwise relationships among population, murders, murder rate, and population density.
- Population and murders exhibit a strong positive relationship, indicating that counties with larger populations generally report more murders.
- Population density shows only a weak positive relationship with murder rate.
- Most variables are positively skewed, as observed from the diagonal density plots.
- A few extreme observations are visible, suggesting the presence of outliers in the dataset.

```python
# Cluster Map of Correlation Matrix

correlation = df[selected_columns].corr()

sns.clustermap(
    correlation,
    cmap="coolwarm",
    annot=True,
    figsize=(8,6)
)

plt.show()
```

### Observation

- The cluster map visualizes the correlations among the selected numerical variables along with hierarchical clustering.
- Population and murders exhibit the strongest positive correlation (0.88), indicating that counties with larger populations generally report more murders.
- Population density has a moderate positive relationship with both population and murders.
- Murder rate shows only weak positive correlations with the other variables, suggesting that additional factors influence crime rates.
- The dendrogram groups population and murders into the same cluster, highlighting their close association.

```python
from sklearn.preprocessing import StandardScaler

scaled_df = pd.DataFrame(
    StandardScaler().fit_transform(df[selected_columns]),
    columns=selected_columns
)

plt.figure(figsize=(10,6))

sns.boxplot(data=scaled_df)

plt.title("Standardized Box Plot of Selected Variables")
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

### Observation

- The standardized box plot enables a fair comparison of the distributions of the selected variables.
- All selected variables exhibit outliers, indicating the presence of extreme observations in the dataset.
- Murder rate shows the most extreme standardized outliers, while population, murders, and population density also contain several high-value observations.
- The median values of the standardized variables are centered around zero, as expected after standardization.

### Overall Observation

- Multivariate analysis was performed using a pair plot, correlation cluster map, and standardized box plot.
- The pair plot revealed pairwise relationships among the selected variables and highlighted the presence of skewness and outliers.
- The correlation cluster map showed that population and murders have the strongest positive correlation, while population density has moderate relationships with population and murders.
- The standardized box plot confirmed the presence of outliers across all selected variables and enabled meaningful comparison by removing the effect of different measurement scales.
- Overall, the multivariate analysis provided a comprehensive understanding of the interactions, correlations, distributions, and variability among the selected variables, supporting further statistical analysis and predictive modeling.
