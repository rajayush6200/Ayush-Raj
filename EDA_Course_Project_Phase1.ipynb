# 9. Time Series Analysis

This section performs Time Series Analysis on the County Murders dataset by aggregating the data annually.

```python
# Import libraries required for Time Series Analysis

from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA

print("Time Series Analysis libraries imported successfully.")
```

```python
# Create an annual time series by aggregating the dataset and calculating annual rates

annual_data = df.groupby("year").agg({
    "murders": "sum",
    "arrests": "sum",
    "popul": "sum"
}).reset_index()

annual_data["murdrate"] = (
    annual_data["murders"] / annual_data["popul"]
) * 10000

annual_data["arrestrate"] = (
    annual_data["arrests"] / annual_data["popul"]
) * 10000

annual_data
```

```python
# Display the basic structure of the annual time series dataset

print("Number of Years:", len(annual_data))
print("Starting Year :", annual_data["year"].min())
print("Ending Year   :", annual_data["year"].max())

print("\nAnnual Time Series Data:")
display(annual_data)
```

```python
# Convert the year column into a datetime index for time-based analysis

annual_data["Date"] = pd.to_datetime(annual_data["year"].astype(str), format="%Y")

annual_ts = annual_data.set_index("Date")

print("Time-based index created successfully.")

annual_ts.head()
```

```python
# Set the annual frequency of the time series

annual_ts = annual_ts.asfreq("YS")

print("Time Series Frequency:", annual_ts.index.freq)
print("Number of observations:", len(annual_ts))
print("Start Date:", annual_ts.index.min())
print("End Date:", annual_ts.index.max())
```

```python
# Display the first and last observations of the time series

print("First Five Observations:")
display(annual_ts.head())

print("\nLast Five Observations:")
display(annual_ts.tail())
```

### Observation

- The County Murders dataset contains annual observations from 1980 to 1996.
- The county-level records were grouped by year to construct an annual time series.
- Total murders and arrests were aggregated using summation, while murder rate and arrest rate were summarized using their annual mean.
- A yearly DateTime index was created to support time-based analysis.
- The resulting time series contains 17 annual observations.

```python
# Plot the annual total number of murders over time

plt.figure(figsize=(12,6))

plt.plot(
    annual_ts.index,
    annual_ts["murders"],
    marker="o",
    linewidth=2
)

plt.title("Annual Total Murders (1980–1996)")
plt.xlabel("Year")
plt.ylabel("Total Number of Murders")

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Observation

- The annual murder series shows a clear decline during the early 1980s.
- Murder totals begin increasing again during the late 1980s.
- The series reaches its highest level around the early 1990s.
- After the early-1990s peak, the total number of murders decreases toward 1996.
- Therefore, the series exhibits a long-term changing trend rather than a constant level.

```python
# Plot the autocorrelation function of the annual murder series

plt.figure(figsize=(10,5))

plot_acf(
    annual_ts["murders"],
    lags=7,
    ax=plt.gca()
)

plt.title("Autocorrelation Function of Annual Murders")
plt.xlabel("Lag")
plt.ylabel("Autocorrelation")

plt.tight_layout()
plt.show()
```

```python
# Calculate and display autocorrelation values for selected lags

print("Autocorrelation Values:")

for lag in range(1, 6):
    print(
        f"Lag {lag}: "
        f"{annual_ts['murders'].autocorr(lag=lag):.4f}"
    )
```

### Observation

- The ACF measures the relationship between the annual murder series and its previous values at different time lags.
- Significant autocorrelation at lower lags indicates that observations in nearby years are not completely independent.
- The ACF can therefore provide useful information when selecting time-series models.

```python
# Plot the partial autocorrelation function of the annual murder series

plt.figure(figsize=(10,5))

plot_pacf(
    annual_ts["murders"],
    lags=5,
    method="ywm",
    ax=plt.gca()
)

plt.title("Partial Autocorrelation Function of Annual Murders")
plt.xlabel("Lag")
plt.ylabel("Partial Autocorrelation")

plt.tight_layout()
plt.show()
```

### Observation

- PACF measures the direct relationship between the current observation and a lag after removing the effects of intermediate lags.
- The PACF helps identify potentially useful autoregressive terms when developing an ARIMA model.
- Because the dataset contains only 17 annual observations, only a small number of lags are considered.

```python
# Perform the Augmented Dickey-Fuller test for stationarity

adf_result = adfuller(
    annual_ts["murders"],
    autolag="AIC"
)

print("Augmented Dickey-Fuller Test")
print("-" * 35)

print(f"ADF Statistic : {adf_result[0]:.4f}")
print(f"p-value       : {adf_result[1]:.4f}")
print(f"Used Lags     : {adf_result[2]}")
print(f"Observations  : {adf_result[3]}")

print("\nCritical Values:")

for key, value in adf_result[4].items():
    print(f"{key}: {value:.4f}")
```

```python
# Interpret the Augmented Dickey-Fuller test result

if adf_result[1] < 0.05:
    print("Result: The series is considered stationary at the 5% significance level.")
else:
    print("Result: The series is not considered stationary at the 5% significance level.")
```

```python
# Apply first-order differencing to remove changes in the series level

murder_diff = annual_ts["murders"].diff().dropna()

print("First-order differenced series:")
display(murder_diff)
```

```python
# Visualize the first-order differenced time series

plt.figure(figsize=(12,6))

plt.plot(
    murder_diff.index,
    murder_diff.values,
    marker="o",
    linewidth=2
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title("First-Order Differenced Annual Murder Series")
plt.xlabel("Year")
plt.ylabel("Differenced Murders")

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Perform the Augmented Dickey-Fuller test after first-order differencing

adf_diff = adfuller(
    murder_diff,
    autolag="AIC"
)

print("ADF Test After First-Order Differencing")
print("-" * 45)

print(f"ADF Statistic : {adf_diff[0]:.4f}")
print(f"p-value       : {adf_diff[1]:.4f}")

if adf_diff[1] < 0.05:
    print("Result: The differenced series is stationary at the 5% significance level.")
else:
    print("Result: The differenced series is not stationary at the 5% significance level.")
```

### Observation

- The original annual murder series is not stationary at the 5% significance level based on the Augmented Dickey-Fuller test.
- First-order differencing was applied; however, the differenced series also does not reject the null hypothesis of a unit root at the 5% significance level.
- Therefore, stationarity is not conclusively established from the ADF test alone.
- Since the dataset contains only 17 annual observations, ARIMA model estimation and forecasting should be interpreted cautiously.

```python
# Compare selected ARIMA models using AIC and BIC

candidate_orders = [
    (0, 1, 0),
    (1, 1, 0),
    (0, 1, 1),
    (1, 1, 1)
]

arima_comparison = []

for order in candidate_orders:
    model = ARIMA(
        annual_ts["murders"],
        order=order
    )

    result = model.fit()

    arima_comparison.append({
        "Order": order,
        "AIC": result.aic,
        "BIC": result.bic
    })

arima_comparison = pd.DataFrame(arima_comparison)

display(
    arima_comparison.sort_values("AIC")
)
```

```python
# Select the ARIMA model with the lowest AIC

best_order = tuple(
    arima_comparison.sort_values("AIC").iloc[0]["Order"]
)

print("Selected ARIMA order based on lowest AIC:", best_order)
```

```python
# Fit the selected ARIMA model

arima_model = ARIMA(
    annual_ts["murders"],
    order=best_order
)

arima_result = arima_model.fit()

print(f"ARIMA{best_order} model fitted successfully.")
```

### Observation

- Residuals should ideally fluctuate around zero without a clear systematic pattern.

## 13. K-Means Clustering

K-Means clustering is an unsupervised learning technique that partitions observations into a predefined number of clusters based on similarity. In this analysis, county-level crime and population variables are standardized and used to identify groups of counties with similar characteristics.

```python
# Select numerical variables for K-Means clustering

clustering_variables = [
    "murders",
    "arrests",
    "popul",
    "murdrate",
    "arrestrate"
]

print("Variables Selected for K-Means Clustering:")
print(clustering_variables)
```

```python
# Create the dataset used for K-Means clustering

kmeans_data = df[clustering_variables].copy()

print("Clustering Dataset Shape:")
print(kmeans_data.shape)

kmeans_data.head()
```

```python
# Check for missing values before clustering

print("Missing Values in Clustering Variables:")
print(kmeans_data.isnull().sum())
```

```python
# Standardize the clustering variables

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

kmeans_scaled = scaler.fit_transform(kmeans_data)

kmeans_scaled = pd.DataFrame(
    kmeans_scaled,
    columns=clustering_variables,
    index=kmeans_data.index
)

kmeans_scaled.head()
```

```python
# Verify the standardized variables

standardized_summary = pd.DataFrame({
    "Mean": kmeans_scaled.mean(),
    "Standard Deviation": kmeans_scaled.std()
})

standardized_summary
```

```python
# Import the K-Means clustering algorithm

from sklearn.cluster import KMeans

print("K-Means algorithm imported successfully.")
```

```python
# Calculate inertia values for different numbers of clusters

inertia_values = []

k_values = range(2, 11)

for k in k_values:
    kmeans_model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans_model.fit(kmeans_scaled)
    inertia_values.append(kmeans_model.inertia_)

inertia_values
```

```python
# Plot the elbow curve to examine the appropriate number of clusters

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.title("Elbow Method for K-Means Clustering")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

```python
# Calculate silhouette scores for different numbers of clusters

from sklearn.metrics import silhouette_score

sample_size = min(5000, len(kmeans_scaled))

silhouette_sample = kmeans_scaled.sample(
    n=sample_size,
    random_state=42
)

silhouette_scores = []

for k in k_values:
    kmeans_model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    sample_labels = kmeans_model.fit_predict(silhouette_sample)

    score = silhouette_score(
        silhouette_sample,
        sample_labels
    )

    silhouette_scores.append(score)

silhouette_scores
```

```python
# Plot silhouette scores for different numbers of clusters

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score for Different Numbers of Clusters")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

```python
# Select the number of clusters based on the highest silhouette score

optimal_k = list(k_values)[np.argmax(silhouette_scores)]

print("Selected Number of Clusters:", optimal_k)
print("Highest Silhouette Score:", round(max(silhouette_scores), 4))
```

```python
# Fit the final K-Means model using the selected number of clusters

kmeans_final = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans_final.fit_predict(kmeans_scaled)

print("Number of Clusters:", optimal_k)
print("Final Inertia:", kmeans_final.inertia_)
```

```python
# Add the K-Means cluster labels to the original dataset

df["KMeans_Cluster"] = cluster_labels

print("Cluster Labels Added Successfully.")
print(df["KMeans_Cluster"].value_counts().sort_index())
```

```python
# Display the K-Means cluster centers in standardized space

cluster_centers_scaled = pd.DataFrame(
    kmeans_final.cluster_centers_,
    columns=clustering_variables
)

cluster_centers_scaled.index.name = "Cluster"

cluster_centers_scaled
```

```python
# Convert cluster centers back to the original variable scale

cluster_centers_original = pd.DataFrame(
    scaler.inverse_transform(
        kmeans_final.cluster_centers_
    ),
    columns=clustering_variables
)

cluster_centers_original.index.name = "Cluster"

cluster_centers_original
```

```python
# Calculate the number of observations in each cluster

cluster_sizes = (
    df["KMeans_Cluster"]
    .value_counts()
    .sort_index()
    .to_frame(name="Number of Observations")
)

cluster_sizes["Percentage"] = (
    cluster_sizes["Number of Observations"]
    / len(df)
) * 100

cluster_sizes
```

```python
# Calculate the mean values of clustering variables for each cluster

cluster_profile = (
    df.groupby("KMeans_Cluster")[clustering_variables]
    .mean()
)

cluster_profile
```

```python
# Calculate the median values of clustering variables for each cluster

cluster_median_profile = (
    df.groupby("KMeans_Cluster")[clustering_variables]
    .median()
)

cluster_median_profile
```

```python
# Visualize K-Means clusters using population and murders

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="popul",
    y="murders",
    hue="KMeans_Cluster",
    palette="viridis",
    alpha=0.6
)

plt.title("K-Means Clusters: Population vs Murders")
plt.xlabel("Population")
plt.ylabel("Number of Murders")
plt.legend(title="Cluster")
plt.show()
```

```python
# Visualize K-Means clusters in three dimensions

fig = plt.figure(figsize=(12, 8))

ax = fig.add_subplot(111, projection="3d")

scatter = ax.scatter(
    df["murders"],
    df["arrests"],
    df["popul"],
    c=df["KMeans_Cluster"],
    cmap="viridis",
    alpha=0.6,
    s=20
)

ax.set_title(
    "3-D Visualization of K-Means Clusters"
)

ax.set_xlabel("Number of Murders")
ax.set_ylabel("Number of Arrests")
ax.set_zlabel("Population")

plt.show()
```

```python
# Visualize the cluster profiles using a heatmap

plt.figure(figsize=(10, 6))

sns.heatmap(
    cluster_profile,
    annot=True,
    fmt=".2f",
    cmap="viridis",
    linewidths=0.5
)

plt.title("K-Means Cluster Profile")
plt.xlabel("Variables")
plt.ylabel("Cluster")

plt.show()
```

```python
# Standardize the cluster profile for comparison across variables

cluster_profile_standardized = (
    cluster_profile -
    cluster_profile.mean()
) / cluster_profile.std()

cluster_profile_standardized
```

```python
# Visualize standardized cluster profiles using a heatmap

plt.figure(figsize=(10, 6))

sns.heatmap(
    cluster_profile_standardized,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Standardized K-Means Cluster Profiles")
plt.xlabel("Variables")
plt.ylabel("Cluster")

plt.show()
```

```python
# Display the main results from K-Means clustering

print("K-Means Clustering Completed")

print("\nNumber of Clusters:", optimal_k)

print("\nFinal Inertia:")
print(round(kmeans_final.inertia_, 4))

print("\nCluster Sizes:")
print(cluster_sizes)

print("\nCluster Mean Profile:")
print(cluster_profile.round(2))
```

```python
# Display the main observation from K-Means clustering

print("Observation:")
print(
    "K-Means clustering groups counties into clusters based on "
    "similarity in murders, arrests, population, murder rate, "
    "and arrest rate. Standardization was applied before clustering "
    "to prevent variables with larger numerical scales from dominating "
    "the distance calculations. The elbow method and silhouette analysis "
    "were used to examine the appropriate number of clusters, followed "
    "by cluster profiling and visualization."
)
```

### Conclusion

K-Means clustering was applied to the County Murders dataset after standardizing the selected numerical variables. The elbow method and silhouette analysis were used to examine the cluster structure, and the resulting clusters were analyzed using cluster sizes, cluster profiles, heatmaps, and two-dimensional and three-dimensional visualizations.

## 14. Hierarchical Clustering

Hierarchical clustering is an unsupervised clustering technique that builds a hierarchy of clusters based on the similarity between observations. In this analysis, Euclidean distance and Ward's method are used to group a representative sample of counties based on standardized crime and population characteristics.

```python
# Select a representative sample and prepare data for hierarchical clustering

hierarchical_sample_size = 2000

hierarchical_sample = df.sample(
    n=hierarchical_sample_size,
    random_state=42
).copy()

hierarchical_data = hierarchical_sample[clustering_variables]

hierarchical_scaler = StandardScaler()

hierarchical_scaled = hierarchical_scaler.fit_transform(
    hierarchical_data
)

print("Original dataset size:", df.shape[0])
print("Hierarchical clustering sample size:", hierarchical_scaled.shape[0])
print("Number of variables:", hierarchical_scaled.shape[1])
```

```python
# Check for missing values in the hierarchical clustering data

print("Missing values in hierarchical clustering data:")
print(hierarchical_data.isnull().sum())
```

```python
# Calculate the Euclidean distance between sampled observations

from scipy.spatial.distance import pdist

hierarchical_distance = pdist(
    hierarchical_scaled,
    metric="euclidean"
)

print("Number of pairwise Euclidean distances:", len(hierarchical_distance))
print("First 10 Euclidean distances:")
print(hierarchical_distance[:10])
```

```python
# Perform hierarchical clustering using Ward's method

from scipy.cluster.hierarchy import linkage

hierarchical_linkage = linkage(
    hierarchical_scaled,
    method="ward",
    metric="euclidean"
)

print("Hierarchical clustering completed using Ward's method.")
print("Linkage matrix shape:", hierarchical_linkage.shape)
```

```python
# Plot the hierarchical clustering dendrogram

from scipy.cluster.hierarchy import dendrogram

plt.figure(figsize=(14, 6))

dendrogram(
    hierarchical_linkage,
    truncate_mode="lastp",
    p=30,
    leaf_rotation=90,
    leaf_font_size=9
)

plt.title("Hierarchical Clustering Dendrogram - Ward's Method")
plt.xlabel("Clustered Observations")
plt.ylabel("Distance")
plt.show()
```

```python
# Create three hierarchical clusters using the Ward linkage results

from scipy.cluster.hierarchy import fcluster

hierarchical_labels = fcluster(
    hierarchical_linkage,
    t=3,
    criterion="maxclust"
)

print("Number of hierarchical clusters:", len(np.unique(hierarchical_labels)))
print("First 20 cluster assignments:")
print(hierarchical_labels[:20])
```

```python
# Count the number of observations in each hierarchical cluster

hierarchical_cluster_counts = pd.Series(
    hierarchical_labels
).value_counts().sort_index()

print("Number of observations in each cluster:")
print(hierarchical_cluster_counts)
```

```python
# Calculate the percentage of observations in each hierarchical cluster

hierarchical_cluster_percentages = (
    hierarchical_cluster_counts / len(hierarchical_labels)
) * 100

hierarchical_cluster_summary = pd.DataFrame({
    "Cluster": hierarchical_cluster_counts.index,
    "Count": hierarchical_cluster_counts.values,
    "Percentage": hierarchical_cluster_percentages.values
})

hierarchical_cluster_summary
```

```python
# Add hierarchical cluster assignments to the sampled observations

hierarchical_sample["Hierarchical_Cluster"] = hierarchical_labels

print("Hierarchical cluster assignments added to the sampled data.")
hierarchical_sample[
    ["Hierarchical_Cluster"]
].head()
```

```python
# Visualize hierarchical clusters using murders and arrests

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=hierarchical_sample,
    x="murders",
    y="arrests",
    hue="Hierarchical_Cluster",
    palette="viridis",
    alpha=0.7
)

plt.title("Hierarchical Clusters: Murders vs Arrests")
plt.xlabel("Murders")
plt.ylabel("Arrests")
plt.legend(title="Cluster")
plt.show()
```

```python
# Visualize hierarchical clusters using population and murders

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=hierarchical_sample,
    x="popul",
    y="murders",
    hue="Hierarchical_Cluster",
    palette="viridis",
    alpha=0.7
)

plt.title("Hierarchical Clusters: Population vs Murders")
plt.xlabel("Population")
plt.ylabel("Murders")
plt.legend(title="Cluster")
plt.show()
```

```python
# Compare hierarchical cluster profiles using a heatmap

plt.figure(figsize=(10, 5))

sns.heatmap(
    hierarchical_mean_profiles,
    annot=True,
    fmt=".2f",
    cmap="viridis"
)

plt.title("Hierarchical Cluster Mean Profiles")
plt.xlabel("Variables")
plt.ylabel("Hierarchical Cluster")
plt.show()
```
